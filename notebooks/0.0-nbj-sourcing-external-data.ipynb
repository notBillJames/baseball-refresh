```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
zg = pd.read_csv('../data/external/zac_gallen_fg_report.csv')

plt.plot(zg["Season"], zg["ERA-"], marker="o")
plt.xlabel("Season")
plt.ylabel("ERA-")
plt.title("Zac Gallen's ERA- Over Time")
plt.xticks(zg["Season"])
plt.grid(True)
plt.show()
```
